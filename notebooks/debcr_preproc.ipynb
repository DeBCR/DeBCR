# DeBCR API tutorial
## Preprocess raw data for DeBCR training/prediction

This notebook shows how to normalize and patch raw microscopy data for DeBCR training/prediction.

```python
import debcr
```

### Load raw data

Set file path to your actual raw input data (look docs for supported formats).

```python
data_filepath = '/path/to/load/data/raw.tif'
data_filepath
```

Let's load data and check its dimentions and min/max values.

```python
data_raw = debcr.data.load(data_filepath)
data_raw.shape, data_raw.min(), data_raw.max()
```

Let's also visually check how raw data looks like.

```python
debcr.data.show([data_raw], slices=[50,70,90], cmap='gray', transpose=True)
```

### Normalize raw data

For DeBCR we recommend to use percentiles to normalize data, which is the default.

```python
data_norm = debcr.data.normalize(data_raw, by_perc=True, pmin=0.1, pmax=99.9)
data_norm.min(), data_norm.max()
```

```python
debcr.data.show([data_norm], slices=[50,70,90], cmap='gray', transpose=True)
```

### Crop patches from normalized data

To ensure full signal coevrage and achieve seamless stitching of patches after prediction, we recommend to crop patches with some overlap, which is the default.

```python
data_ptch = debcr.data.crop(data_norm, patch_size=128, overlap=(0.5, 0.5))
data_ptch.shape
```

The 'dry_run=*True*' instead of cropped patches returns amount of patches (in total and along each axis) according to the passed parameter.

We need to know this for stitching them back after prediction.

```python
patch_cnt, patch_num = debcr.data.crop(data_norm, patch_size=128, overlap=(0.5, 0.5), dry_run=True)
patch_cnt, patch_num
```

```python
debcr.data.show([data_ptch], slices=[-1,-1,-1,-1], cmap='gray', transpose=True)
```

Let's make a dry-run stitch to verify that stack dimensions after assembly would match the data before cropping

```python
asmbl_cnt, asmbl_size = debcr.data.stitch(data_ptch, patch_num, overlap=(0.5, 0.5), dry_run=True)
asmbl_cnt, asmbl_size
```

```python
data_norm.shape
```

### Save pre-processed data

Finally, let's save pre-processed data.

```python
data_prep_filepath = '/path/to/save/data/preproc.npy'
data_prep_filepath
```

```python
debcr.data.write(data_prep_filepath, data=data_ptch)
```

The pre-processed data is now suitable for running predictions or training DeBCR model (see the tutorial links
on the GitHub).
