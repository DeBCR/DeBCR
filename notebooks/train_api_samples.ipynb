# DeBCR API tutorial
## Predict from sample data/weights

```python
import os
import numpy as np

import debcr
```

### Setup paths to samples (data and weights)

```python
# Available datasets: FM_CARE_2D, FM_CARE_3D, CF_STED_FActin, WF_SIM_S.aureus
dataset_name = 'FM_CARE_2D'
dataset_path = f'/path/to/debcr_samples/{dataset_name}'

dataset_path
```

### Load training/validation data

```python
# Choose subset from test, train, val
data_dirpath = f'{dataset_path}/data'
data_dirpath
```

```python
train_data_filepath = f'{data_dirpath}/{dataset_name}_train.npz'
val_data_filepath = f'{data_dirpath}/{dataset_name}_val.npz'
train_data_filepath, val_data_filepath
```

```python
data_train = debcr.data.load(train_data_filepath)
```

```python
# npz can store multiple arrays as .files
data_train.files
```

```python
# We provide two arrays: 
# gt  - groundtruth data
# low - low-exposure data to be deblurred

print("Input data shape: ", data_train["low"].shape)
print("Groundtruth data shape: ", data_train["gt"].shape)
```

```python
import random
from debcr._core.show_utils import subShow

slice_num = random.randint(0, data_train["gt"].shape[0]-1)
subShow(data_train["gt"][slice_num].squeeze(), data_train["low"][slice_num].squeeze())
print(slice_num)
```

```python
data_val = debcr.data.load(val_data_filepath)
```

```python
import random
from debcr._core.show_utils import subShow

slice_num = random.randint(0, data_val["gt"].shape[0]-1)
subShow(data_val["gt"][slice_num].squeeze(), data_val["low"][slice_num].squeeze())
print(slice_num)
```

### Loading config

```python
# to load default config for training
config = debcr.config.load()
config
```

### Train model on training data

```python
my_model = debcr.model.train(data_train, data_val, config)
```

### Test model on validation data

```python
data_pred = debcr.model.predict(my_model, data_val["low"])
data_pred.shape
```

```python
import random
from debcr._core.show_utils import subShow3

slice_num = random.randint(0, data_val["gt"].shape[0]-1)
subShow3(data_val["gt"][slice_num].squeeze(), data_pred[slice_num], data_val["low"][slice_num].squeeze())
print(slice_num)
```
